```python
# (2026-27 start)

import requests

rawText = """
KL - Agr, Consumer & Env Sciences
KY - Applied Health Sciences
LD - Armed Forces
NQ - Auxiliary Units
LT - Carle Illinois Medicine
LN - Center Innov in Teach Learn
NA - Chancellor
KT - College of Media
NU - Div Intercollegiate Athletics
KW - Division of Exploratory Studies
KN - Education
MY - Energy Services
NN - Facilities
KR - Fine & Applied Arts
KM - Gies College of Business
KS - Graduate College
KP - Grainger Engineering
LQ - Illinois International
KU - Law
KV - Liberal Arts & Sciences
NP - OVCRI Institutes
NS - Ofc VC Inst Advancement
NB - Provost & Executive VC
LF - Public Safety
LP - School of Information Sciences
LG - School of Labor & Empl. Rel.
LL - School of Social Work
NT - Technology Services
LR - University Library
NW - VC Administrative Strategy
NL - VC Finance and Administration
NJ - VC Student Affairs
NE - VC for Research and Innovation
LC - Veterinary Medicine
NV - Vice Chancellor for Access
"""

collegeCodes = []
collegeNames = []
for line in rawText.split("\n"):
  if len(line) > 2:
    code, name = line.split(" - ")

    collegeCodes.append(code)
    collegeNames.append(name)

print(collegeCodes)
print(collegeNames)
```

```python
import io

def processPage(rawHTML, collegeCode, collegeName):
  dfs = []
  h3s = rawHTML.split("<h3")
  for index in range(1, len(h3s)):
    h3_split = h3s[index]

    h3_full = "<h3" + h3_split
    h3_split = h3_split[ h3_split.find(">") + 1 : h3_split.find("</h3>") ]
    departmentName = h3_split.split("-")[1].strip()
    print(departmentName)

    df = pd.read_html(io.StringIO(h3_full))[0]
    df["Department Name"] = departmentName
    df["College Code"] = collegeCode
    df["College Name"] = collegeName
    dfs.append(df)

  if len(dfs) > 0:
    return pd.concat(dfs)
  else:
    return None
```

```python
# Test script
rawHTML = requests.get(f"https://www.trustees.uillinois.edu/trustees/resources/26-27-Graybook/KP.html").text
rawHTML

processPage(rawHTML, "KP", "Grainger")
```

```python
from time import sleep

dfs = []
for i in range(len(collegeCodes)):
  collegeCode = collegeCodes[i]
  collegeName = collegeNames[i]
  print(f"=== {collegeName} ===")
  rawHTML = requests.get(f"https://www.trustees.uillinois.edu/trustees/resources/26-27-Graybook/{collegeCode}.html").text

  result = processPage(rawHTML, collegeCode, collegeName)
  if type(result) is not None:
    dfs.append(result)
  sleep(5)

df = pd.concat(dfs)
```

```python
df["Proposed Salary"] = df["Proposed Salary"].replace('[\$,]', '', regex=True).astype(float)
```

```python
df["Department Name"] = df["Department Name"].str.replace('&amp;', '&')
```

```python
df["College Name"] = df["College Name"].replace('&amp;', '&')
```

```python
def process(dfE):
  df_jobs = dfE[ dfE["Job Title"] != "Employee Total for All Jobs" ]
  
  jobTitles = list(df_jobs["Job Title"].unique())
  depts = list(df_jobs["Department Name"].unique())
  
  highest_salary_row = df_jobs.nlargest(1, "Proposed Salary")
  jobTitles.remove(highest_salary_row["Job Title"].values[0])
  depts.remove(highest_salary_row["Department Name"].values[0])

  if len(jobTitles) == 0:
    jobTitles = ""
  if len(depts) == 0:
    depts = ""

  return {
    "Employee Name": highest_salary_row["Employee Name"].values[0],

    "Primary Job Title": highest_salary_row["Job Title"].values[0],
    "Department Name": highest_salary_row["Department Name"].values[0],
    "College Code": highest_salary_row["College Code"].values[0],
    "College Name": highest_salary_row["College Name"].values[0],

    "Present FTE": dfE["Present FTE"].max(),
    "Proposed FTE": dfE["Proposed FTE"].max(),
    "Present Salary": dfE["Present Salary"].max(),
    "Proposed Salary": dfE["Proposed Salary"].max(),

    "Tenure": highest_salary_row["Tenure"].values[0],
    "Empl Class": highest_salary_row["Empl Class"].values[0],

    "Secondary Job Titles": str(jobTitles),
    "Secondary Department Names": str(depts),
  }


d = []
for name in df["Employee Name"].unique():
  dfSingle = df[ df["Employee Name"] == name ]

  dfAllJobs = dfSingle[ dfSingle["Job Title"] == "Employee Total for All Jobs" ]

  if len(dfAllJobs) > 1:
    values = dfAllJobs["Present Salary"].values

    val = values[0]
    identicalName = False
    for v in values:
      if v != val:
        identicalName = True

    if identicalName:
      if name == "Kumar, Praveen":
        indexFilter = [154, 155]
      else:
        print(f"!!! {name}")
        
      df1 = dfSingle[ dfSingle.index.isin(indexFilter) ]
      df2 = dfSingle[ ~dfSingle.index.isin(indexFilter) ]

      d.append(process(df1))
      d.append(process(df2))
    else:
      d.append(process(dfSingle))
  else:
    d.append(process(dfSingle))
```

```python
df_CLEAN = pd.DataFrame(d)
```
